```python
#filling all the null values with median value of same category. Like null values in clothing category will filled by median of clothing category. Likewise Footwear category null values will filled by the median of footwear category only. 
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# firstly convert all the lower and uppercase character into lowercase
# secondly replace all the spaces between words with underscore'_'
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create new column. name: 'age_group'
```

```python
labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create new column. name: 'purchase_frequency_days'

#how often customer purchase the product. It is in text "weekly, monthly or yearly", It's hard to deal in text for analysis. So, we will convert text into number.
```

```python
frequency_mapping={
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

server = r"PRAKHAR-PANDEY\MSSQL"
database = "customer_behaviour"

connection_string = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    "mssql+pyodbc:///?odbc_connect=" + quote_plus(connection_string)
)
```

```python
with engine.connect() as connection:
    print("Connected successfully!")
```

```python
#Load DataFrame into SQL Server
#Let's create a table called customer_data:
```

```python
table_name = "customer_data"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False
)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
